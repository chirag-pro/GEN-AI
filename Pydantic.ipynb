```python
class Product(BaseModel):
  title:str=Field(max_length=18)
  des:str=Field(min_length=10)
  price:int=Field(gt=70,le=10000000)
```

```python
pro=Product(title="iphone17",des="this is amazing cell phone use for many cases",price=120000)
```

```python
class Pickup(BaseModel):
  title: str = Field(max_length=25)
  price: int = Field(gt=100, le=10000)
  items: List[Product]
  comment: str = Field(default="No Comment")
```
