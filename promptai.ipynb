```python
import langchain
```

```python
from langchain_groq import ChatGroq
```

```python
llm=ChatGroq(model="openai/gpt-oss-120b")
```

```python
prompts = [
    {"role":"user", "content":"Hello, I am Chirag."},
    {"role":"ai", "content":'Hello, Chirag! How can I assist you today?'},
    {"role":"user", "content":"What is My name and  my location ?"}
]

res = llm.invoke(prompts)
```

```python
history = [] ## [{"role":"user", "content":"Who is pm of india ?"}]
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("Good Bye 👋")
        break

    history.append({"role":"user", "content":query})
    print("User: ", query)


    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI: ", res.content, "\n")
```
