```python
import langchain
```

```python
from langchain_groq import ChatGroq
```

```python
llm=ChatGroq(
        api_key='',
    model='openai/gpt-oss-120b'
)
```

```python
llm.invoke('hi').content
```

```python
query= input('how was the problen today')
res= llm.invoke(query).content
print(res)
```

```python
query=''
while query!= 'end':
  query= input('ask anything')
  if query== 'end':
    break
  res= llm.invoke(query).content
  print(res)
```

```python
hist=[]
query=''
while query!= 'end':
  query= input('ask anything')
  hist.append({'role': 'user', 'content': query})
  if query== 'end':
    break
  res= llm.invoke(hist).content
  hist.append({'role': 'ai', 'content': res})
  print(res)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt= [ {'role': 'system', 'content': 'you are a Javascript developer'},
         {'role': 'user', 'content': 'what is highorder function'}
         ]
```

```python
print(llm.invoke(prompt).content)
```

```python
promt2= [{'role':'system', 'content': 'you are a english to {language} translator'},
         {'role': 'user', 'content': '{query}'}]
```

```python
P2= ChatPromptTemplate.from_messages(promt2)
```

```python
P2
```

```python
P3=P2.invoke({'language': 'french', 'query': 'hello how are you?'})
```

```python
llm.invoke(P3).content
```

```python
llm.invoke(P2.invoke({'language': 'hindi', 'query': 'hello how are you?'})).content
```

```python
def transform(txt):
  return txt.content.upper()
```

```python
chain = P2 | llm | transform
```

```python
chain.invoke({'language': 'English', 'query': 'नमस्ते, आप कैसे हैं?'})
```
